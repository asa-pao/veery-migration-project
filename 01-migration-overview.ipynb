# Mapping migration

### An introduction to vector data operations

The Veery thrush (*Catharus fuscescens*) migrates each year between nesting sites in North America and South America. Veeries are small birds, and encountering a hurricane during their migration would be disasterous. However, Ornithologist Christopher Heckscher has found in tracking their migration that their breeding patterns and migration timing can predicting the severity of hurricane season as accurately as meterological models [(Heckscher 2018)](https://doi.org/10.1038/s41598-018-28302-3)!

1. What are some reasons that animals might migrate?
- Breeding is the central reason I know. In Oregon, where I'm from, the salmon move upstream from the ocean to fresh water to spawn.
- Food is another one. Animals migrate to find better water or grazing (typically heading north). We see whales head north along the coast in the summer. Although humpback whales don't find much feed where they breed.

2. How might climate change affect animal migrations?
- In North America, species have been migrating earlier than usual.
- Along with this, they have been mating earlier.
- Climate change also indirectly affects migration via food. With changes in food seasons and availability, animals migrate differently.
